# Workflows

How the JBPHH drinking water archive pipeline works, one diagram per workflow. Stepped
diagrams have **Back / Next** buttons and a caption for each step; the others are images.

**To view:** open this notebook in VS Code, choose this folder's `.venv` as the kernel
(**Select Kernel**), then **Run All**. Diagrams of the archive load from the Shared Drive
named by `ARCHIVE_ROOT` in `.env`; diagrams of this code load from `docs/diagrams/`.

**To add or change a stepped diagram:** edit or add a spec in `docs/diagrams/specs/`,
rebuild it with `uv run archive diagrams build` (see the README), then add a section below.

```python
"""Find the diagram folders and show diagrams. Run this cell first."""

import os
from pathlib import Path

from IPython.display import HTML, Image, display

REPO_DIAGRAMS = Path("docs") / "diagrams"


def archive_diagrams() -> Path:
    """The archive's docs/diagrams folder, from ARCHIVE_ROOT in the environment or in .env."""
    root = os.environ.get("ARCHIVE_ROOT")
    if not root and Path(".env").exists():
        for line in Path(".env").read_text(encoding="utf-8").splitlines():
            if line.startswith("ARCHIVE_ROOT="):
                root = line.split("=", 1)[1].strip().strip('"')
    if not root:
        raise RuntimeError("Set ARCHIVE_ROOT in .env (see .env.example).")
    return Path(root) / "docs" / "diagrams"


def show(folder: Path, name: str) -> None:
    """Show a diagram: the stepped version if there is one, otherwise its PNG."""
    stepped = folder / f"{name}.embed.html"
    if stepped.exists():
        display(HTML(filename=stepped))
    else:
        display(Image(filename=folder / f"{name}.png"))
```

## The archive at a glance

Where documents come from, how they are collected, where they are kept, and who uses them.

```python
show(archive_diagrams(), "archive-big-picture")
```
