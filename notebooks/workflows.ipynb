# Workflows

How the JBPHH drinking water reports archive pipeline works, one diagram per workflow. Stepped
diagrams have **Back / Next** buttons and a caption for each step; the others are images.

**To view:** open this notebook in VS Code, choose the repo's `.venv` as the kernel
(**Select Kernel**), then **Run All**. All diagrams load from the repo's `docs/diagrams/`.

**To add or change a stepped diagram:** edit or add a spec in `docs/diagrams/specs/`,
rebuild it with `uv run archive diagrams build` (see the README), then add a section below.

```python
"""Show diagrams from docs/diagrams/. Run this cell first."""

from IPython.display import HTML, Image, display

from jbphh_archive.diagrams.paths import DIAGRAMS_FOLDER, embed_path, png_path


def show(name: str) -> None:
    """Show a diagram: the stepped version if there is one, otherwise its PNG."""
    stepped = embed_path(DIAGRAMS_FOLDER, name)
    if stepped.exists():
        display(HTML(filename=stepped))
    else:
        display(Image(filename=png_path(DIAGRAMS_FOLDER, name)))
```
